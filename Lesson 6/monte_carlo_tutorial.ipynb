# Monte Carlo Method


## 1. Random Numbers in Python
Python's `random` module and `numpy.random` provide tools for generating random numbers, which are essential for Monte Carlo simulations.

### Basic Random Number Generation

```python
import random
import numpy as np

# Set seed for reproducibility
random.seed(42)
np.random.seed(42)

# Uniform random numbers
print(random.random())  # Single float between 0 and 1
print(random.random())  # Single float between 0 and 1
print(np.random.rand(3))  # Array of 3 floats between 0 and 1

# Random integers
print(random.randint(1, 10))  # Integer between 1 and 10
print(np.random.randint(1, 10, size=3))  # Array of 3 integers
```

```python
# Normal distribution
normal_samples = np.random.normal(loc=0, scale=1, size=1000)  # Mean=0, std=1

# Uniform distribution
uniform_samples = np.random.uniform(low=-1, high=1, size=1000)  # Between -1 and 1
```

## 2. What is the Monte Carlo Method?
The Monte Carlo method uses random sampling to solve problems that may be deterministic or probabilistic. 
It approximates solutions by simulating many **random trials**.

### Key Steps
1. Define the problem and its domain.
2. Generate random inputs based on a probability distribution.
3. Perform computations on these inputs.
4. Aggregate results to estimate the solution.

## 3. Example 1: Estimating Pi
A classic Monte Carlo example is estimating the value of π by simulating random points in a square and circle.

### Problem
- Consider a unit square (side length 2) centered at (0,0).
- Inscribe a unit circle (radius 1) inside it.
- The ratio of points falling inside the circle to total points approximates the ratio of the circle's area to the square's area: $(π * r^2) / (2r)^2 = π/4$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set seed
np.random.seed(42)

# Number of points
n = 10000

# Generate random points
x = np.random.uniform(-1, 1, n)
y = np.random.uniform(-1, 1, n)

# Check if points lie inside the unit circle
inside_circle = x**2 + y**2 <= 1

# Estimate pi
pi_estimate = 4 * np.sum(inside_circle) / n
print(f"Estimated π: {pi_estimate}")

# # Visualize
# plt.figure(figsize=(6, 6))
# plt.scatter(x[inside_circle], y[inside_circle], c='b', s=1, label='Inside Circle')
# plt.scatter(x[~inside_circle], y[~inside_circle], c='r', s=1, label='Outside Circle')
# plt.gca().set_aspect('equal')
# plt.legend()
# plt.show()
```

### Explanation
- Random points are generated uniformly in the square.
- Points where `x² + y² ≤ 1` lie inside the unit circle.
- The ratio of points inside the circle to total points, multiplied by 4, estimates π.
- As `n` increases, the estimate approaches the true value of π (≈3.14159).

## 4. Example 2: Numerical Integration
Monte Carlo can approximate integrals, especially in high dimensions.

### Problem
Estimate the integral: $\int_{0}^{1} x² dx$.
- Analytically, the result is `1/3 ≈ 0.3333`.

```python
import numpy as np

# Set seed
np.random.seed(42)

# Number of samples
n = 10000000

# Generate random points
x = np.random.uniform(0, 1, n)
y = np.random.uniform(0, 1, n)

# Evaluate function

under_curve = y < x**2 
# Estimate integral

integral_estimate = np.sum(under_curve) / n * 1
print(f"Estimated integral: {integral_estimate}")
```
